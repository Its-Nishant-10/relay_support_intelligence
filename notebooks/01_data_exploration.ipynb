# Support Ticket Dataset Exploration
This notebook reproduces the exploratory analysis for the public `Prady06/customer-support-tickets` dataset. The dataset card describes it as synthetic IT ticket data. Run `python scripts/download_dataset.py` first.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
import seaborn as sns
sys.path.append(str(Path.cwd().parent))
from triage.data import load_public_support_tickets
frame = load_public_support_tickets()
frame.head()
```

```python
frame['text_length'] = (frame['subject'].fillna('').astype(str) + ' ' + frame['body'].fillna('').astype(str)).str.len()
print('Tickets:', len(frame))
print('Duplicate subject/body pairs:', frame.duplicated(subset=['subject', 'body']).sum())
display(frame.isna().sum().sort_values(ascending=False).to_frame('missing'))
display(frame['text_length'].describe().to_frame('text_length'))
```

```python
figures = Path('../reports/figures')
figures.mkdir(parents=True, exist_ok=True)
for column in ['type', 'priority', 'queue', 'language']:
    plt.figure(figsize=(10, 4))
    frame[column].value_counts().head(20).plot(kind='bar', color='#7ea63c')
    plt.title(f'Ticket distribution by {column}')
    plt.tight_layout()
    plt.savefig(figures / f'distribution_{column}.png', dpi=160)
    plt.show()
plt.figure(figsize=(10, 4))
sns.histplot(frame['text_length'], bins=40, color='#275d55')
plt.title('Ticket intake text length distribution')
plt.tight_layout()
plt.savefig(figures / 'text_length_distribution.png', dpi=160)
plt.show()
```
